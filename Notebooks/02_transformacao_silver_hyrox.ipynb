```python
# Importar as funções necessárias do PySpark
from pyspark.sql import functions as F
from pyspark.sql.types import TimestampType


# Ler a tabela bruta da camada Bronze
df_bronze = spark.table("workspace.default.hyrox_results")


# Identificar automaticamente todas as colunas importadas como timestamp
# Não precisamos digitar manualmente o nome de cada tempo
colunas_timestamp = [
    campo.name
    for campo in df_bronze.schema.fields
    if isinstance(campo.dataType, TimestampType)
]


# Criar o DataFrame da camada Silver a partir dos dados brutos
df_silver = df_bronze


# Converter cada duração timestamp para segundos e minutos
for nome_coluna in colunas_timestamp:

    # Criar uma coluna com o tempo total em segundos
    df_silver = df_silver.withColumn(
        f"{nome_coluna}_segundos",
        (
            F.hour(F.col(nome_coluna)) * 3600
            + F.minute(F.col(nome_coluna)) * 60
            + F.second(F.col(nome_coluna))
        )
    )

    # Criar uma coluna com o tempo total em minutos
    df_silver = df_silver.withColumn(
        f"{nome_coluna}_minutos",
        F.round(F.col(f"{nome_coluna}_segundos") / 60, 2)
    )


# Salvar a tabela tratada na camada Silver
df_silver.write.mode("overwrite").saveAsTable(
    "workspace.default.silver_hyrox_resultados"
)


# Informar quais colunas foram convertidas
print("Tabela Silver criada com sucesso.")
print("Colunas convertidas:")
print(colunas_timestamp)
```

```python
# Importar as funções necessárias do PySpark
from pyspark.sql import functions as F


# Ler a tabela tratada da camada Silver
df_silver = spark.table("workspace.default.silver_hyrox_resultados")


# Selecionar somente registros com tempo total válido
# Valores nulos ou iguais a zero não devem entrar nos benchmarks
df_validos = df_silver.filter(
    (F.col("total_time_segundos").isNotNull()) &
    (F.col("total_time_segundos") > 0) &
    (F.col("gender").isNotNull()) &
    (F.col("age_group").isNotNull()) &
    (F.col("division").isNotNull())
)


# Agrupar os atletas por gênero, faixa etária e divisão
# e calcular as estatísticas de desempenho
df_gold_benchmarks = df_validos.groupBy(
    "gender",
    "age_group",
    "division"
).agg(
    F.count("*").alias("quantidade_atletas"),

    F.round(
        F.min("total_time_segundos"),
        2
    ).alias("melhor_tempo_segundos"),

    F.round(
        F.avg("total_time_segundos"),
        2
    ).alias("tempo_medio_segundos"),

    F.round(
        F.expr("percentile_approx(total_time_segundos, 0.50)"),
        2
    ).alias("mediana_tempo_segundos"),

    F.round(
        F.expr("percentile_approx(total_time_segundos, 0.25)"),
        2
    ).alias("percentil_25_tempo_segundos"),

    F.round(
        F.expr("percentile_approx(total_time_segundos, 0.75)"),
        2
    ).alias("percentil_75_tempo_segundos"),

    F.round(
        F.max("total_time_segundos"),
        2
    ).alias("pior_tempo_segundos")
)


# Criar versões dos indicadores em minutos
df_gold_benchmarks = (
    df_gold_benchmarks
    .withColumn(
        "melhor_tempo_minutos",
        F.round(F.col("melhor_tempo_segundos") / 60, 2)
    )
    .withColumn(
        "tempo_medio_minutos",
        F.round(F.col("tempo_medio_segundos") / 60, 2)
    )
    .withColumn(
        "mediana_tempo_minutos",
        F.round(F.col("mediana_tempo_segundos") / 60, 2)
    )
    .withColumn(
        "percentil_25_tempo_minutos",
        F.round(F.col("percentil_25_tempo_segundos") / 60, 2)
    )
    .withColumn(
        "percentil_75_tempo_minutos",
        F.round(F.col("percentil_75_tempo_segundos") / 60, 2)
    )
    .withColumn(
        "pior_tempo_minutos",
        F.round(F.col("pior_tempo_segundos") / 60, 2)
    )
)


# Salvar os benchmarks na camada Gold
df_gold_benchmarks.write.mode("overwrite").saveAsTable(
    "workspace.default.gold_hyrox_benchmarks"
)


# Confirmar a criação da tabela
print("Tabela Gold criada com sucesso.")
print("Total de grupos:", df_gold_benchmarks.count())
```
